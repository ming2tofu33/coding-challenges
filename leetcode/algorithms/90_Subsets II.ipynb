```python
from typing import List

class Solution:
    def subsetsWithDup(self, nums: List[int]) -> List[List[int]]:
        nums.sort()
        result = []
        path = []

        def backtrack(start: int):
            result.append(path[:])

            for i in range(start, len(nums)):
                # 같은 depth에서 같은 숫자를 또 선택하면 중복 subset이 생김
                if i > start and nums[i] == nums[i - 1]:
                    continue

                path.append(nums[i])
                backtrack(i + 1)
                path.pop()

        backtrack(0)
        return result
```

```python
Solution().subsetsWithDup(nums = [1,2,2])
```

```python
Solution().subsetsWithDup(nums = [0])
```
