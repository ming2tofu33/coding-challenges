```python
from typing import List
import math

class Solution:
    def constructRectangle(self, area: int) -> List[int]:
        w = math.isqrt(area)

        while area % w != 0:
            w -= 1

        l = area // w
        return [l, w]
```

```python
Solution().constructRectangle(4)
```

```python
Solution().constructRectangle(37)
```
